```python
data = [(1, "A", 100), (2, "B", 200), (3, "A", 300)]
df = spark.createDataFrame(data, ["id", "name", "amount"])
```

```python
df.write.mode("overwrite").parquet(
  "dbfs:/Volumes/dev/bronze/my_volume/orders_parquet"
)
```

```python
df_read = spark.read.parquet(
  "dbfs:/Volumes/dev/bronze/my_volume/orders_parquet"
)

df_read.show()
```

```python
df.write.mode("overwrite").csv(
  "dbfs:/Volumes/dev/bronze/my_volume/orders_csv",
  header=True
)
```

```python
df_csv = spark.read.option("header", True).csv(
  "dbfs:/Volumes/dev/bronze/my_volume/orders_csv"
)
df_csv.show()
```

```python
df.write.format("delta").save(
  "dbfs:/Volumes/dev/bronze/my_volume/orders_delta"
)
```

```python
df_delta = spark.read.format("delta").load(
  "dbfs:/Volumes/dev/bronze/my_volume/orders_delta"
)
df_delta.show()
```
